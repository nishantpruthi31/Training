```
# create pandas DataFrame
import pandas as pd

dfp = pd.DataFrame(
    [(1, "Steve"), (2, "Bob")],
    columns=["id", "name"])
dfp
```

```
# create Snowpark DataFrame from pandas DataFrame
from snowflake.snowpark.context import get_active_session

session = get_active_session()
df = session.create_dataframe(dfp)
df
```

```
# save pandas DataFrame to temporary/transient table
df = session.write_pandas(dfp,
    "pandas_table",
    auto_create_table=True,
    table_type="temp")
df
```

```
# Snowpark DataFrame to pandas DataFrame
dfp = df.to_pandas()
dfp

# Snowpark DataFrame to pandas DataFrame batches (not all loaded in memory)
dfp = df.to_pandas_batches()
dfp
```

```
# Snowpark DataFrame to pandas in Snowpark DataFrame (different API!)
dfsp = df.to_snowpark_pandas()
dfsp
```
